```python
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

from google.colab import files

uploaded = files.upload()

df = pd.read_csv("car_advertisement dataset.csv", encoding='latin1')

print("First 5 Rows:")
print(df.head())

print("\nColumns:")
print(df.columns)
```

```python
df = df.dropna()
```

```python
X = df.drop("price", axis=1)
y = df["price"]

print("\nShape of X:", X.shape)
print("Shape of Y:", y.shape)
```

```python
categorical_columns = X.select_dtypes(include=["object"]).columns

X = pd.get_dummies(X, columns=categorical_columns)

X = X.astype(int)

print("\nEncoded Data:")
print(X.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

print("\nScaled Features:")
print(X_scaled.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.33,
    random_state=42
)

print("\nTraining and Testing Shapes:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel Intercept:")
print(model.intercept_)

print("\nModel Coefficients:")
print(model.coef_)
```

```python
y_pred = model.predict(X_test)

print("\nFirst 10 Predicted Values:")
print(y_pred[:10])

print("\nFirst 10 Actual Values:")
print(y_test.iloc[:10].values)
```

```python
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print("\nR2 Score:", r2)
print("Mean Absolute Error:", mae)
```

```python
joblib.dump(model, "LR_car_advertisement.pkl")

joblib.dump(scaler, "scaler_car_advertisement.pkl")

joblib.dump(list(X.columns), "columns_car_advertisement.pkl")

print("\nModel and preprocessing files saved successfully!")
```
